# AULA 4 — POPULAÇÃO

Análise da população dos municípios e estados entre 2010 e 2022.


```python
import pandas as pd
from pathlib import Path

PASTA = Path.cwd()

arquivos_excel = list(PASTA.glob('*.xlsx'))
if not arquivos_excel:
    raise FileNotFoundError('Nenhum arquivo .xlsx encontrado. Coloque a tabela anexada pelo professor na pasta do projeto.')

path = arquivos_excel[0]
print(f'Arquivo utilizado: {path.name}')
```

```python
populacao_bruto = pd.read_excel(path, header=None)
display(populacao_bruto.head())
```

```python
populacao_total = populacao_bruto.iloc[3:, 1:].copy()
populacao_total.columns = ['UF', 'cod_uf', 'cod_munic', 'municipio', 'pop_2010_sinopse', 'pop_2010', 'pop_2022']

populacao_total = populacao_total.dropna(subset=['UF', 'municipio']).copy()
populacao_total['pop_2010'] = pd.to_numeric(populacao_total['pop_2010'], errors='coerce')
populacao_total['pop_2022'] = pd.to_numeric(populacao_total['pop_2022'], errors='coerce')
populacao_total = populacao_total.dropna(subset=['pop_2010', 'pop_2022']).copy()

display(populacao_total.head())
```

```python
populacao_estado = (
    populacao_total.groupby('UF', as_index=False)[['pop_2010', 'pop_2022']]
    .sum()
)

populacao_estado['crescimento'] = populacao_estado['pop_2022'] - populacao_estado['pop_2010']
populacao_estado = populacao_estado.sort_values('crescimento', ascending=False)

display(populacao_estado)
```

```python
arquivo_estado = PASTA / 'populacao_estado_crescimento.csv'
populacao_estado.to_csv(arquivo_estado, index=False, sep=';', encoding='utf-8-sig')
print(f'Salvo em: {arquivo_estado}')
```

```python
populacao_municipio = populacao_total[
    ['UF', 'cod_uf', 'cod_munic', 'municipio', 'pop_2010', 'pop_2022']
].copy()

populacao_municipio['crescimento'] = (
    populacao_municipio['pop_2022'] - populacao_municipio['pop_2010']
)
populacao_municipio = populacao_municipio.sort_values('crescimento', ascending=False)

display(populacao_municipio)
```

```python
arquivo_municipio = PASTA / 'populacao_municipio_crescimento.csv'
populacao_municipio.to_csv(arquivo_municipio, index=False, sep=';', encoding='utf-8-sig')
print(f'Salvo em: {arquivo_municipio}')
```
